```python
%sql
use gsynergy.default;


select c.fsclwk_id,pos_site_id, f.sku_id,price_substate_id , type

,sum(f.sales_units) as total_units
from gold_fact_transactions f 
inner join gold_dim_clnd c on f.fscldt_id = c.fscldt_id
left join  gold_dim_possite p on f.pos_site_id = p.site_id
left join gold_dim_product pr on f.sku_id = pr.sku_id
left join gold_dim_state s on f.price_substate_id = s.substate_id
where c.fsclwk_id = '201801' and f.sku_id = '0310920000' 
group by c.fsclwk_id,pos_site_id, f.sku_id,price_substate_id , type
order by c.fsclwk_id
```
