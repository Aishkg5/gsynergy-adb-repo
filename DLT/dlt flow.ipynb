```python
import dlt
from pyspark.sql.functions import date_format, current_timestamp
```

## BRONZE

```python


@dlt.table(
    name ='bronze_price_state',
    schema="""
    state_id STRING NOT NULL PRIMARY KEY,
    state_label STRING,
    _rescued_data STRING,
    file_processed_date STRING
    """
)
def state():
    df = spark.readStream.format("cloudFiles")\
      .option("cloudFiles.format", "parquet")\
      .load(f"{location}/price_state/")\
      .withColumn('file_processed_date',date_format(current_timestamp(),'yyyy-MM-dd HH:mm:ss'))
    return df
```

## SILVER

```python
dlt.create_streaming_table(
	name = 'silver_price_state'
)
dlt.apply_changes(
	target = 'silver_price_state',
	source = 'bronze_price_state',
	keys = ['state_id'],
	stored_as_scd_type = 1,
	sequence_by = '1',
    except_column_list = ['_rescued_data','file_processed_date']
)
```

```python
dlt.create_streaming_table(
	name = 'silver_price_substate'
)
dlt.apply_changes(
	target = 'silver_price_substate',
	source = 'bronze_price_substate',
	keys = ['state_id'],
	stored_as_scd_type = 1,
	sequence_by = '1',
    except_column_list = ['_rescued_data','file_processed_date']
)
```
